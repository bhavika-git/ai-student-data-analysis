```python
import matplotlib.pyplot as plt
burnout_data=df.groupby('Burnout_Risk_Level')['Weekly_GenAI_Hours'].mean()
burnout_data.plot(kind='bar',color=['red','green','orange'])
plt.title('Average AI Hours by Burnout Risk Level')
plt.xlabel('Burnout Risk Level')
plt.ylabel('Average Weekly AI Hours')
```

```python
gpa_data=df.groupby('Major_Category')['Post_Semester_GPA'].mean()
gpa_data.plot(kind='bar',color='blue')
plt.title('Average GPA by Major Category')
plt.xlabel('Major Category')
plt.ylabel('Average Post Semester GPA')
plt.xticks(rotation=40)
plt.show()
```

```python
An=df.groupby('Anxiety_Level_During_Exams')['Post_Semester_GPA'].mean()
An.plot(kind='line',color='purple',marker='o')
plt.title('Average GPA by Anxiety Level During Exams')
plt.xlabel('Anxiety Level During Exams')
plt.ylabel('Average Post Semester GPA')
plt.show()
```

```python
## My Insights from AI Student Dataset

1. Students with high burnout use ~15 hrs of AI/week vs ~4.6 hrs for low burnout students
2. STEM students have the highest average GPA (3.36) after the semester
3. All majors perform similarly in GPA — around 3.3 to 3.4
4. High AI usage correlates with high burnout ,that is when AI usage increases then the risk
   of burnout also increases — but causation is unclear meaning that this does not mean that
   one causes the other
5. Students with the highest level of exam anxiety (level 10) show a sharp decline in their
   average post-semester GPA (dropping to around 3.26)s, indicating that extreme exam
   stress negatively impacts academic performance
```
